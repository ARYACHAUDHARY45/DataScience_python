```python
#we want to find title of all books
```

```python
articles = soup.select("article.product_pod")
```

```python
items=[]
for article in articles :
    title =  article.find("h3").find("a")["title"]
    price= article.select_one("p.price_color").text.split("£")[1]
    rating_element=article.select_one("p.star-rating")
    ratings =rating_element['class'][1]
    items.append([title,price,ratings])
```

```python
import pandas as pd
df= pd.DataFrame(items,columns=["Book Title","Price","Ratings"])
df
```

```python
df.to_csv("data.csv",index=False)
```
